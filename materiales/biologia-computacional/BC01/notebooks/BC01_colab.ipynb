## 4. Verificación con checksums

Comparamos los checksums SHA-256 para garantizar integridad.

```python
import subprocess
import hashlib
import os

# Verificar mini.tsv
result = subprocess.run(['sha256sum', 'datos/mini.tsv'], capture_output=True, text=True)
actual_hash = result.stdout.split()[0]
with open('datos/SHA256SUMS') as f:
    # La línea de mini.tsv
    for line in f:
        expected = line.split()[0]
        print(f'mini.tsv: SHA-256 = {actual_hash}')
        print(f'          Esperado = {expected}')
        match = actual_hash == expected
        print(f'          Estado: {"OK ✅" if match else "FALLO ❌"}')

# Verificar MANIFEST
result_m = subprocess.run(['sha256sum', '-c', 'datos/MANIFEST.sha256'], capture_output=True, text=True)
lines_out = result_m.stdout.strip().split('\n')
passed = sum(1 for l in lines_out if ': OK' in l)
failed = sum(1 for l in lines_out if ': FAILED' in l)
print(f'\nMANIFEST: {passed} OK, {failed} fallidos')
```

## 7. Creación del manifiesto de entorno

Genera un manifiesto reproducible con la información del entorno.

```python
import os, platform, datetime, hashlib, subprocess

manifiesto = []
manifiesto.append('=== Manifiesto de entorno — BC01 ===')
manifiesto.append(f'Fecha: {datetime.datetime.utcnow().isoformat()}')
manifiesto.append(f'Plataforma: {platform.platform()}')
manifiesto.append(f'Python: {platform.python_version()}')
manifiesto.append(f'Shell: {os.environ.get("SHELL", "N/A")}')
manifiesto.append(f'PWD: {os.getcwd()}')
manifiesto.append('')

for f in ['datos/mini.tsv', 'datos/caso_s01.tsv']:
    with open(f, 'rb') as fh:
        h = hashlib.sha256(fh.read()).hexdigest()
    manifiesto.append(f'{f}: {h}')

print('\n'.join(manifiesto))
```
